Exploratory Data Analysis of Bordeaux real estate dataset. The code for generating the plots is in this notebook.

```python
"""
Data Analysys for Bordeaux real estate dataset
Generate plots and save to results/figures/
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

def load_cleaned_data(filepath='data/processed/bordeaux_clean.csv'):
    """Load clean dataset"""
    if "__file__" in globals():
        file_parent = Path(__file__).resolve().parent
        if file_parent.name == "notebooks":
            base_dir = file_parent.parent
        else:
            base_dir = file_parent
    else:
        base_dir = Path.cwd()
        if base_dir.name == "notebooks":
            base_dir = base_dir.parent
    csv_path = base_dir / filepath
    df = pd.read_csv(csv_path)
    print(f"Loaded cleaned data with shape: {df.shape}")
    return df

def save_plot(fig, filename, dpi=300):
    """Save a matplotlib figure to results/figures/"""
    # Base directory that work in both script and notebook contexte
    if "__file__" in globals():
        file_parent = Path(__file__).resolve().parent
        if file_parent.name == "notebooks":
            base_dir = file_parent.parent
        else:
            base_dir = file_parent
    else:
        base_dir = Path.cwd()
        if base_dir.name == "notebooks":
            base_dir = base_dir.parent
    fig_dir = base_dir / 'results' / 'figures'
    fig_dir.mkdir(parents=True, exist_ok=True)
    fig_path = fig_dir / filename
    fig.savefig(fig_path, dpi=dpi, bbox_inches='tight')
    print(f"Saved figure: {fig_path}")
    plt.close(fig) # Free memory

def plot_price_distribution(df):
    """Histogram of price per m²"""
    fig, ax = plt.subplots()
    sns.histplot(df['prix_m2'], bins=50, kde=True, ax=ax, color='skyblue')
    ax.set_title('Distribution of Price per Square Meter (€/m²)')
    ax.set_xlabel('Price per m² (€)')
    ax.set_ylabel('Frequency')
    save_plot(fig, 'price_distribution.png')

def plot_price_by_property_type(df):
    """Plot boxplot of price per m² by property type"""
    fig, ax = plt.subplots()
    # Order by median price for better visualisation
    order = df.groupby('type_local')['prix_m2'].median().sort_values().index
    sns.boxplot(x='type_local', y='prix_m2', data=df, order=order, ax=ax, palette='Set2')
    ax.set_title('Price per Square Meter by Property Type')
    ax.set_xlabel('Property Type')
    ax.set_ylabel('Price per m² (€)')
    ax.tick_params(axis='x', rotation=45)
    save_plot(fig, 'price_by_property_type.png')

def plot_price_by_postal_code(df, top_n=10):
    """Plot bar chart of average price per square meter for top N postal codes"""
    # Calculate average price per postal code
    avg_price_by_postal = df.groupby('code_postal')['prix_m2'].mean().sort_values(ascending=False)
    # Take top N
    top_postal = avg_price_by_postal.head(top_n)

    fig, ax = plt.subplots()
    sns.barplot(x=top_postal.index, y=top_postal.values, ax=ax, palette='viridis')
    ax.set_title(f'Top {top_n} Postal Codes by Average Price per m²')
    ax.set_xlabel('Postal Code')
    ax.set_ylabel('Average Price per m² (€)')
    ax.tick_params(axis='x', rotation=45)
    save_plot(fig, f'price_by_top{top_n}_postal_codes.png')

def plot_correlation_heatmap(df):
    """Plot heatmap of correlations between key numerical variables."""
    # Correlation
    numerical_cols = ['prix_m2', 'surface_reelle_bati', 'nombre_pieces_principales',
                      'latitude', 'longitude', 'valeur_fonciere']
    # Existence
    numerical_cols = [col for col in numerical_cols if col in df.columns]
    corr_df = df[numerical_cols].corr()

    fig, ax = plt.subplots()
    sns.heatmap(corr_df, annot=True, fmt=".2f", cmap='coolwarm', center=0,
                square=True, linewidths=0.5, cbar_kws={"shrink": .8}, ax=ax)
    ax.set_title('Correlation Heatmap of Key Variables')
    save_plot(fig, 'correlation_heatmap.png')

def main():
    """Main EDA function."""
    print("Starting Exploratory Data Analysis...")
    df = load_cleaned_data()

    # Plots
    plot_price_distribution(df)
    plot_price_by_property_type(df)
    plot_price_by_postal_code(df, top_n=10)
    plot_correlation_heatmap(df)

    print("EDA completed. Figures saved to results/figures/")

if __name__ == "__main__":
    main()
```
